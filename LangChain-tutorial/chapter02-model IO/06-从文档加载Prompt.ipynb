# 从物理磁盘的yaml中读取prompt数据

```python
from langchain_core.prompts import load_prompt
from dotenv import load_dotenv

load_dotenv()
prompt = load_prompt("./prompt.yaml", encoding="utf-8")
# print(prompt)
print(prompt.format(name="年轻人", what="滑稽"))
```

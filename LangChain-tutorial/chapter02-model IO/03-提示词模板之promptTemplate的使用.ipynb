## 1. PromptTemplate 如何获取实例
方式1：使用构造方法的方式

```python
from langchain.chains.qa_with_sources.stuff_prompt import template
from langchain_core.prompts import PromptTemplate

# 1、创建promptTemplate实例
# 参数中必须要知名：input_variables 、template
prompt_template =  PromptTemplate(
    template = "你是一个{role},你的名字是{name}",
    input_variables = ["role","name"],
)

# 2、填充实例中的变量。现在暂且使用format()
prompt = prompt_template.format(role="人工智能专家",name="小明")
print(prompt)
```

# 两种特殊结构的使用（部分提示词模板的使用、组合提示词的使用）
## 2.1 部分提示此模板的使用
举例： 在PromptTemplate的构造方法或from_template()的方法内使用partial_variables来进行设置

```python
from langchain_core.prompts import PromptTemplate

# 1、创建promptTemplate实例
# 参数中必须要知名：input_variables 、template
prompt_template =  PromptTemplate(
    template = "你是一个{role},你的名字是{name}",
    input_variables = ["role","name"],
    partial_variables={"role":"人工智能专家"}

)

# 2、填充实例中的变量。现在暂且使用format()
prompt = prompt_template.format(name="小明")
print(prompt)
```

```python
from langchain_core.prompts import PromptTemplate

# 1、创建promptTemplate实例
# 参数中必须要知名：input_variables 、template
prompt_template =  PromptTemplate.from_template(
    template = "你是一个{role},你的名字是{name}",
    partial_variables={"role":"人工智能专家"}
)

# 2、填充实例中的变量。现在暂且使用format()
prompt = prompt_template.format(name="小明")
print(prompt)
```

## 调用方法 partial()
partial()调用完以后，不会对调用这个模板对象产生影响；它会返回一个新的模板，因该使用这个新的模板

```python
from langchain_core.prompts import PromptTemplate

# 1、创建promptTemplate实例
# 参数中必须要知名：input_variables 、template
prompt_template =  PromptTemplate(
    template = "你是一个{role},你的名字是{name}",
    input_variables = ["role","name"],
)
# 该函数返回了一个新的模板
new_prompt_template = prompt_template.partial(role="是个专家")
# 2、填充实例中的变量。现在暂且使用format()
prompt = new_prompt_template.format(name="小明")
print(prompt)
```

```python
from langchain_core.prompts import PromptTemplate

# 1、创建promptTemplate实例
# 参数中必须要知名：input_variables 、template
prompt_template =  PromptTemplate(
    template = "你是一个{role},你的名字是{name}",
    input_variables = ["role","name"],
).partial(role="是个专家")
# 2、填充实例中的变量。现在暂且使用format()
prompt = prompt_template.format(name="小明")
print(prompt)
```

## 组合提示词的使用

```python
from langchain_core.prompts import PromptTemplate
template = (
PromptTemplate.from_template(template="Tell me a joke about {topic}")
+ ", make it funny"
+ "\n\nand in {language}"
)
prompt = template.format(topic="sports", language="spanish")
print(prompt)
```

## 给变量赋值的两种方式：format(),invoke()
format()，返回值为字符串类型；
invoke()，返回值为PromptValue类型，接着调用to_string()返回字符串。

```python
from langchain_core.prompts import PromptTemplate

# 1、创建promptTemplate实例
# 参数中必须要知名：input_variables 、template
prompt_template =  PromptTemplate.from_template(
    template = "你是一个{role},你的名字是{name}",
).partial(role="是个专家")

# 2、填充实例中的变量。现在暂且使用format()
prompt_format = prompt_template.format(name="小明")
print(prompt_format)
print(type(prompt_format))
prompt_invoke = prompt_template.invoke(input={"name":"小明"})
print(prompt_invoke)
print(type(prompt_invoke))
```

## 结合大模型的使用

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
import os
import dotenv

# 加载环境变量
dotenv.load_dotenv(
    # 在工程根目录只保留一个 .env，所有模块都引用它。这是最不容易出错的方式。
    override=True,  # 如果加载.env文件出现了同名环境变脸，会进行覆盖
    # dotenv_path=".env",  # 可以只当需要加载的路径
)
chat_model = ChatOpenAI(
    api_key=os.getenv("SILICON_FLOW_API_KEY"),
    base_url=os.getenv("SILICON_FLOW_BASE_URL"),
    model="Qwen/Qwen3-8B",  # 确保模型名称准确
    temperature=0.7,
)

# 生成提示词模板
prompt_template = PromptTemplate.from_template(
template = "请评价{product}的优缺点，包括{aspect1}和{aspect2}。"
)
# 给模板的变量赋值
prompt = prompt_template.format(product="电脑",aspect1="性能",aspect2="电池")

res = chat_model.invoke(prompt)

print(res)
```

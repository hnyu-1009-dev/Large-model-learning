# LangChain调用本地大模型

### 第二步 在Ollam下部署大模型：
### 第三步 在LangChain中调用Ollama下的大模型

```python
import os

# ==========================================
# 🌟 杀手锏：绕过系统代理防御 502 报错
# 强制让 Python 在访问本地地址时不经过任何 VPN 或代理软件
# ==========================================
os.environ["NO_PROXY"] = "localhost,127.0.0.1"
os.environ["no_proxy"] = "localhost,127.0.0.1"

from langchain_ollama import ChatOllama
from langchain_core.messages import HumanMessage
import dotenv

# 加载环境变量
dotenv.load_dotenv(override=True)

# 明确指定 base_url 为 127.0.0.1（比 localhost 更底层，进一步防止解析错误）
llm = ChatOllama(
    model="qwen2.5:0.5b",
    base_url="http://127.0.0.1:11434" 
)

messages = [
    HumanMessage(content="你好，请介绍一下你自己！"),
]

print("🚀 正在呼叫本地大模型 qwen2.5:0.5b ...")
response = llm.invoke(messages)

print("\n✅ 成功回复：")
print(response.content)
```

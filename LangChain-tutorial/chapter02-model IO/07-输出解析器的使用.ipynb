# 1、字符串输出解析器StrOutputParser

```python
from langchain_openai import ChatOpenAI
import os
import dotenv
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    HumanMessage,
    SystemMessage,
    AIMessage,
)
from langchain_core.prompts import (
    FewShotChatMessagePromptTemplate,
    PromptTemplate,
    ChatPromptTemplate,
    SystemMessagePromptTemplate,
    HumanMessagePromptTemplate,
    AIMessagePromptTemplate,
    FewShotPromptTemplate,
)

# 加载环境变量
dotenv.load_dotenv(
    # 在工程根目录只保留一个 .env，所有模块都引用它。这是最不容易出错的方式。
    override=True,  # 如果加载.env文件出现了同名环境变脸，会进行覆盖
    # dotenv_path=".env",  # 可以只当需要加载的路径
)


# 1.提供大模型
chat_model = ChatOpenAI(
    api_key=os.getenv("SILICON_FLOW_API_KEY"),
    base_url=os.getenv("SILICON_FLOW_BASE_URL"),
    model="Qwen/Qwen3-8B",  # 确保模型名称准确
)

# 2.调用大模型
response = chat_model.invoke("什么是大语言模型？")

# 3.如何获取一个字符串的输出结果呢？
# 方式1：自己调用输出结果的contect

# 方式2：使用StrOutputParser
parser = StrOutputParser()
str_response = parser.invoke(response)
print(type(str_response))
print(str_response)
```

## JSON解析器JsonOutputParser

```python
from langchain_openai import ChatOpenAI
import os
import dotenv
from langchain_core.output_parsers import (
    StrOutputParser,
    JsonOutputParser,
)
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    HumanMessage,
    SystemMessage,
    AIMessage,
)
from langchain_core.prompts import (
    FewShotChatMessagePromptTemplate,
    PromptTemplate,
    ChatPromptTemplate,
    SystemMessagePromptTemplate,
    HumanMessagePromptTemplate,
    AIMessagePromptTemplate,
    FewShotPromptTemplate,
)

# 加载环境变量
dotenv.load_dotenv(
    # 在工程根目录只保留一个 .env，所有模块都引用它。这是最不容易出错的方式。
    override=True,  # 如果加载.env文件出现了同名环境变脸，会进行覆盖
    # dotenv_path=".env",  # 可以只当需要加载的路径
)


# 1.提供大模型
chat_model = ChatOpenAI(
    api_key=os.getenv("SILICON_FLOW_API_KEY"),
    base_url=os.getenv("SILICON_FLOW_BASE_URL"),
    model="Qwen/Qwen3-8B",  # 确保模型名称准确
)
chat_prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个靠谱的{role}"),
        ("human", "{question}"),
    ]
)

# 2.调用大模型
response = chat_model.invoke(
    chat_prompt_template.format_messages(
        role="人工智能专家",
        question="人工智能用英文怎么说？问题用q表示，答案用a表示，返回一个JSON格式",
    )
)

# 3.如何获取一个字符串的输出结果呢？
# 方式1：自己调用输出结果的contect

# 方式2：使用StrOutputParser
parser = JsonOutputParser()
Json_response = parser.invoke(response)
print(type(Json_response))
print(Json_response)
```

```python
from langchain_openai import ChatOpenAI
import os
import dotenv
from langchain_core.output_parsers import (
    StrOutputParser,
    JsonOutputParser,
)
from langchain_openai import ChatOpenAI
from langchain_core.messages import (
    HumanMessage,
    SystemMessage,
    AIMessage,
)
from langchain_core.prompts import (
    FewShotChatMessagePromptTemplate,
    PromptTemplate,
    ChatPromptTemplate,
    SystemMessagePromptTemplate,
    HumanMessagePromptTemplate,
    AIMessagePromptTemplate,
    FewShotPromptTemplate,
)

# 加载环境变量
dotenv.load_dotenv(
    # 在工程根目录只保留一个 .env，所有模块都引用它。这是最不容易出错的方式。
    override=True,  # 如果加载.env文件出现了同名环境变脸，会进行覆盖
    # dotenv_path=".env",  # 可以只当需要加载的路径
)


# 1.提供大模型
chat_model = ChatOpenAI(
    api_key=os.getenv("SILICON_FLOW_API_KEY"),
    base_url=os.getenv("SILICON_FLOW_BASE_URL"),
    model="Qwen/Qwen3-8B",  # 确保模型名称准确
)
chat_prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "你是一个靠谱的{role}"),
        ("human", "{question}"),
    ]
)

# 2.调用大模型
response = chat_model.invoke(
    chat_prompt_template.format_messages(
        role="人工智能专家",
        question="人工智能用英文怎么说？问题用q表示，答案用a表示，返回一个JSON格式",
    )
)
parser = JsonOutputParser()

prompt_template = PromptTemplate.from_template(
    template="回答用户的查询\n 满足的格式为{format_instructions}\n 问题为{question}\n",
    partial_variables={
        "format_instructions": parser.get_format_instructions(),
    },
)

prompt = prompt_template.invoke(
    input={"question": "告诉我一个笑话"},
)

response = chat_model.invoke(prompt)
print(response)

json_result = parser.invoke(response)
print(json_result)
```

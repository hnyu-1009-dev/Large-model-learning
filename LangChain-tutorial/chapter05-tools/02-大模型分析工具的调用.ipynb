# 定义工具

```python
# ==============================================================================
# 🛠️ 模块一：工具的创建与参数解析
# ==============================================================================
from langchain_core.tools import tool, StructuredTool

# ------------------------------------------------------------------------------
# 方式 1：最基础的 @tool 装饰器 (懒人最爱)
# 特点：全自动！默认把【函数名】作为工具名，把【"""文档字符串"""】作为工具描述。
# ------------------------------------------------------------------------------
@tool
def add_number(a: int, b: int) -> int:
    """两个整数相加"""  # 💡 这句话极其重要！大模型就是看这句话来决定用不用这个工具的。
    return a + b

print("--- 方式 1: 基础 @tool ---")
print(f"自动提取的名字: {add_number.name}")           # 输出: add_number
print(f"自动提取的描述: {add_number.description}")    # 输出: 两个整数相加
print(f"自动提取的参数: {add_number.args}")           # 输出: {'a': {'title': 'A', 'type': 'integer'}, ...}

# ------------------------------------------------------------------------------
# 方式 2：带参数的 @tool 装饰器 (企业级常用)
# 特点：可以强行覆盖函数原本的名字和描述，还能开启高阶开关。
# ------------------------------------------------------------------------------
@tool(
    name_or_callable="add_two_number", # 强行改名
    description="当用户要求将两个数字相加时调用此工具", # 强行覆盖描述（给大模型看的）
    return_direct=True # 🌟 核心开关：设为 True 时，工具执行完直接把结果扔给用户，大模型不再废话总结。
)
def add_number_advanced(a: int, b: int) -> int:
    """这里的注释会被上面的 description 覆盖"""
    return a + b

print("\n--- 方式 2: 高阶 @tool(...) ---")
print(f"覆盖后的名字: {add_number_advanced.name}")
print(f"覆盖后的描述: {add_number_advanced.description}")
print(f"直返开关状态: {add_number_advanced.return_direct}")

# ------------------------------------------------------------------------------
# 方式 3：StructuredTool.from_function (封装第三方源码的终极杀招)
# 特点：如果你用的是别人写好的库，你不能去改别人的源码加 @tool 装饰器，
#      这时候就可以用 StructuredTool 把别人的普通函数强行“包”成工具！
# ------------------------------------------------------------------------------
def search_function(query: str):
    return f"【模拟联网结果】关于 {query}，LangChain 是个好框架。"

search_tool = StructuredTool.from_function(
    func=search_function,       # 绑定真正的普通函数
    name="SearchTool",          # 取个名字
    description="当需要查询最新事件或事实时使用此工具" 
)

print("\n--- 方式 3: StructuredTool ---")
print(f"工具名称: {search_tool.name}")

# ==============================================================================
# 🎯 知识点：万物皆可 invoke！
# 现在这些工具已经不是普通的 Python 函数了，它们是 LangChain 的 Tool 对象。
# 执行它们统一使用 .invoke() 方法，并传入【字典】格式的参数。
# ==============================================================================
res = add_number_advanced.invoke({"a": 10, "b": 20})
print(f"\n执行结果: {res}")
```

```python
import os
import json
import dotenv
from typing import List
from langchain_core.pydantic_v1 import BaseModel, Field
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_openai import ChatOpenAI

# ==============================================================================
# ⚙️ 阶段 1：大模型初始化参数
# ==============================================================================
dotenv.load_dotenv(override=True)

chat_model = ChatOpenAI(
    api_key=os.getenv("SILICON_FLOW_API_KEY"),
    base_url=os.getenv("SILICON_FLOW_BASE_URL"),
    # 参数: model
    # 作用: 指定要调用的具体大模型名称。
    # 细节: 现代 LangChain 依赖这个名字来决定底层使用哪种方言（比如 Qwen、GPT-4 要求的 JSON 格式略有不同，底层会根据名字自动适配）。
    model="Qwen/Qwen3-8B",
    # 参数: temperature
    # 取值范围: 通常是 0.0 到 2.0（根据模型厂商有所不同）。
    # 作用: 控制模型输出的“随机性”和“创造力”。
    # 铁律: 只要你的模型需要调用工具（尤其是需要输出严格 JSON 参数时），务必设为 0！否则它可能会创造出不存在的参数键名导致崩溃。
    temperature=0,
)


# ==============================================================================
# 🧠 阶段 2：Pydantic 参数校验模型 (给大模型看的说明书)
# ==============================================================================
class ScheduleMeetingInput(BaseModel):
    # 参数: description (存在于 Field 中)
    # 作用: 【极度重要】这绝对不是写给程序员看的注释！这是直接喂给大模型看的“提取指令”。
    # 机制: 大模型会阅读这段话，从用户的模糊提问中抠出符合该描述的字眼，填到 title 变量里。
    title: str = Field(description="会议的主题或核心目的，尽量精简")

    # 类型提示: List[str]
    # 作用: 强制要求大模型输出一个字符串数组。如果用户说“叫上张三和李四”，大模型会自动转为 ["张三", "李四"]。
    participants: List[str] = Field(description="必须参加此会议的人员姓名或部门列表")

    # 参数: default (存在于 Field 中)
    # 作用: 默认值兜底。如果用户的话里根本没提开多久的会，大模型可以不提取此参数，程序会自动使用 default 指定的 30。
    duration: int = Field(
        default=30,
        description="会议的预计时长（单位：分钟），如果用户未提及则使用默认值",
    )

    is_urgent: bool = Field(
        default=False,
        description="是否是紧急、马上、立刻要开的会议。是设为True，否则False",
    )


# ==============================================================================
# 🛠️ 阶段 3：工具函数的声明与参数
# ==============================================================================
# 参数: args_schema
# 作用: 将上面写好的 Pydantic 类强行绑定给这个工具。
# 机制: 逼迫大模型在调用本工具前，必须通过 ScheduleMeetingInput 的严苛检查。如果不加这个参数，大模型就只能看到基础的 title: str。
@tool(args_schema=ScheduleMeetingInput)
def schedule_meeting(
    # 这里的参数名 (title, participants 等) 必须和上面 BaseModel 里的属性名【完全一模一样】！
    title: str,
    participants: List[str],
    duration: int,
    is_urgent: bool,
) -> str:
    """这是一个用于在公司日历系统上预定会议、安排日程的专用工具。"""

    # --- 模拟本地 Python 业务防错逻辑 ---
    if len(participants) == 0:
        return "预定失败：必须提供至少一名参会人员名单！请向用户追问参会人。"

    if duration > 120:
        return "预定失败：单次会议时长上限为 120 分钟，请缩短会议时间。"

    urgency_tag = "🚨 [紧急]" if is_urgent else "📅 [普通]"
    return f"【系统通知】已成功执行！{urgency_tag} 会议【{title}】预定成功，时长 {duration} 分钟。参会人：{', '.join(participants)}。"


# ==============================================================================
# 🔗 阶段 4：工具绑定参数
# ==============================================================================
tools = [schedule_meeting]

# 参数: tools
# 作用: 接收一个包含工具对象的列表（List）。
# 机制: bind_tools 会在底层将这个列表里的所有 Python 函数，翻译成 OpenAI 格式的 JSON Schema，并附加在下一次大模型的请求中。
model_with_tools = chat_model.bind_tools(tools)
# model_with_tools它是一个“被武装过的大模型”。


# ==============================================================================
# 🚀 阶段 5：执行闭环与 ToolMessage 参数详解
# ==============================================================================
print("\n💬 开始请求大模型...")
messages = [
    HumanMessage(
        content="立刻帮我建个紧急会，讨论一下双十一大促复盘，只拉张三和项目组长就行，大概开1个小时。"
    )
]

# 参数: input / messages
# 作用: 触发模型思考。传入完整的对话历史列表。
response = model_with_tools.invoke(messages)
# response 是一个它返回的是一个 AIMessage 对象
print("response = model_with_tools.invoke(messages):", response)
# 必须把大模型这次“决定调用工具”的消息原封不动存进列表，因为这是上下文的一部分！
messages.append(response)

if response.tool_calls:
    for tool_call in response.tool_calls:
        print("tool_call",tool_call)
        tool_name = tool_call["name"]

        # 参数: tool_call["args"]
        # 本质: 这是一个完美的 Python 字典，大模型已经按照 ScheduleMeetingInput 的要求把数据全填好了！
        tool_args = tool_call["args"]

        # 参数: tool_call["id"]
        # 本质: 这是一个长串字符串（如 call_abc123...），是大模型随机生成的任务流水号。
        tool_call_id = tool_call["id"]

        print(
            f"✅ 大模型成功解析参数:\n{json.dumps(tool_args, ensure_ascii=False, indent=2)}"
        )

        if tool_name == "schedule_meeting":
            # 参数: tool_args (字典)
            # 作用: 使用 kwargs (关键字参数) 的形式，把大模型提取的字典直接传给 Python 函数运行。
            tool_result = schedule_meeting.invoke(tool_args)

            # 🌟 ToolMessage 参数终极解析 🌟
            tool_msg = ToolMessage(
                # 参数: tool_call_id (必填！)
                # 作用: 对账/销账单号。大模型之前下发指令时带了 id，你现在汇报结果必须原样把 id 传回去，否则大模型会抛出异常拒绝接收。
                tool_call_id=tool_call_id,
                # 参数: name (可选但强烈推荐)
                # 作用: 明确告诉大模型，这是哪个工具产生的返回结果，防止并发调用多个工具时混乱。
                name=tool_name,
                # 参数: content (必填！)
                # 作用: Python 函数在本地跑完后的返回结果。
                # 铁律: 必须转化为字符串类型 (str)！即使你原本的工具返回了列表或字典，也要用 json.dumps() 转成字符串填在这里。
                content=str(tool_result),
            )
            # 把结果凭证追加进历史记录
            messages.append(tool_msg)

print("\n⏳ 大模型正在根据工具结果生成友好回复...")
# 带着最新追加的 ToolMessage 再次请求大模型，完成闭环汇报
final_response = model_with_tools.invoke(messages)
print(f"\n🤖 最终回复:\n{final_response.content}")
```

Tool 的本质：它就是一座桥梁，通过描述（description）让大模型知道何时该用它，通过参数校验（args_schema）逼迫大模型输出格式化参数，然后你在本地用 Python 执行真正的逻辑。

新老语法的鸿沟：老代码里满屏幕的 functions 和 json.loads 都是为了手动处理数据格式。而现在，一个 .bind_tools() 和一个 .tool_calls 就把底层脏活全包了！

```python
import os
import dotenv
from langchain_community.tools import MoveFileTool
from langchain_core.messages import HumanMessage
from langchain_openai import ChatOpenAI

# 1. 基础配置
# 2. 定义严格的大模型 (调用工具温度务必设为 0)
dotenv.load_dotenv(
    # 在工程根目录只保留一个 .env，所有模块都引用它。这是最不容易出错的方式。
    override=True,  # 如果加载.env文件出现了同名环境变脸，会进行覆盖
    # dotenv_path=".env",  # 可以只当需要加载的路径
)

chat_model = ChatOpenAI(
    api_key=os.getenv("SILICON_FLOW_API_KEY"),
    base_url=os.getenv("SILICON_FLOW_BASE_URL"),
    model="Qwen/Qwen3-8B",  # 确保模型名称准确
)

# 3. 准备工具箱 (我们可以把前面自己写的加法工具，和官方的移动文件工具放在一起)
# 注意：MoveFileTool 需要提供确切的绝对路径
tools = [MoveFileTool()]

# ==============================================================================
# 🚀 步骤 1：大模型意图识别 (工具绑定)
# ==============================================================================
print("🧠 步骤 1: 正在把工具交给大模型，并下达移动文件的指令...")
# 【现代写法】：一键绑定，生成“带手”的新模型
model_with_tools = chat_model.bind_tools(tools)

# 故意给一个明确的指令
messages = [
    HumanMessage(
        content="请帮我把当前目录下的 'abc.txt' 移动到 'C:\\Users\\Desktop\\abc.txt'"
    )
]

# 请求大模型。注意：大模型此时【绝对不会】真的去移动文件，它只是在写“施工图纸”！
response = model_with_tools.invoke(messages)

# ==============================================================================
# 🔍 步骤 2：解析大模型的指令图纸
# ==============================================================================
print("\n🔍 步骤 2: 检查大模型是否决定使用工具...")

# 【现代写法】：直接判断 response.tool_calls 列表
if not response.tool_calls:
    print(f"💬 大模型觉得不需要用工具，纯文本回复: {response.content}")
else:
    print(f"✅ 成功捕获工具调用指令！数量: {len(response.tool_calls)}")

    # ==========================================================================
    # ⚙️ 步骤 3：物理执行工具
    # ==========================================================================
    for tool_call in response.tool_calls:
        tool_name = tool_call["name"]
        tool_args = tool_call["args"]  # 【现代写法】拿出来直接就是完美字典！

        print(f"\n   ▶️ 准备调用的工具: {tool_name}")
        print(f"   ▶️ 提取到的参数: {tool_args}")

        # 寻找匹配的工具并执行
        if tool_name == "move_file":
            print("   🚀 开始在本地真实执行文件移动...")
            executor = MoveFileTool()
            try:
                # 【现代写法】使用 .invoke() 替代旧版的 .run()
                result = executor.invoke(tool_args)
                print(f"   🎉 执行结果: {result}")
            except Exception as e:
                print(f"   ❌ 执行失败 (可能是文件路径不存在): {e}")
```
